### 1st Step-Indexing(document Ingestion)

```python
from youtube_transcript_api import YouTubeTranscriptApi,TranscriptsDisabled

video_id="xAt1xcC6qfM"

try:
    
    trascript=YouTubeTranscriptApi().fetch(video_id=video_id,languages=['en']).to_raw_data()
    #print(trascript)
    cleaned_transcript="".join([chunk['text'] for chunk in trascript])
    with open('trns.txt','w+') as f:
        f.write(cleaned_transcript)
except TranscriptsDisabled:
    print("No trascript available")
```

### 1st Step-Indexing(Embedding generation & store in vector store)

```python
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
embedded_model=HuggingFaceEmbeddings(model="sentence-transformers/all-MiniLM-L6-v2")
```

```python
vector_store=FAISS.from_documents(
    embedding=embedded_model,
    documents=chunk
)
```

```python
vector_store.index_to_docstore_id
```

```python
vector_store.get_by_ids(['45c247b9-aea9-45ca-a348-8c2f4f782e05'])
```

### Step-2 Retriever:

```python
retriever= vector_store.as_retriever(search_type="similarity",search_kwargs={'k':4})
```

```python
retriever
```

```python
retriever.invoke("What is told by Bill Gates here")
```

### 3.Augmentation

```python
from langchain_core.prompts import PromptTemplate
```

```python
template=PromptTemplate(
    template="""You are a helpful assistant.
      Answer ONLY from the provided transcript context.
      If the context is insufficient, just say you don't know.

      {context}
      Question: {question}
    """,
    input_variables=['context','question']
)
```

```python
query="Is Bill Gates talk about modern AI technologies here?"
retriever_docs=retriever.invoke(query)
```

```python
context_text="\n\n".join(doc.page_content for doc in retriever_docs)
```

```python
context_text
```

```python
final_prompt=template.invoke({'context':context_text,'question':query})
```

```python
final_prompt
```

### Step-4 Genaration

```python
from langchain_huggingface import ChatHuggingFace,HuggingFaceEndpoint
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
llm=HuggingFaceEndpoint(
    repo_id="deepseek-ai/DeepSeek-V3.1", 
    task="text-generation"
    
)
model= ChatHuggingFace(llm=llm)
```

```python
answer=model.invoke(final_prompt)
print(answer.content)
```

### Same thing implemented using chain

```python
from langchain_core.runnables import RunnableParallel,RunnablePassthrough,RunnableLambda
from langchain_core.output_parsers import StrOutputParser
```

```python
def retriver_output(retriever_doc):
    context_text="\n\n".join(doc.page_content for doc in retriever_doc)
    return context_text  
```

```python
parser=StrOutputParser()
```

```python
paraller_chain=RunnableParallel(
    {'context':retriever | RunnableLambda(retriver_output),
     'question':RunnablePassthrough()
     }  
)

model_chain= template | model | parser

main_chain=paraller_chain|model_chain
```

```python
output=main_chain.invoke("does Bill Gates talk about AI technologies?")
```

```python
print(output)
```

```python
"""Hmm, the user is asking if Bill Gates discusses AI technologies in the provided transcript. Let me scan the context carefully.

I see several mentions of AI. Gates talks about AI being the "today's equivalent" of what he did in his youth, emphasizing its potential. He also mentions risks and opportunities, like building new companies in AI. There's a part where he advises using AI for creative work and learning, highlighting its educational benefits.

The context clearly shows Gates discussing AI, so the answer should confirm this. I need to ensure the response stays strictly within the provided transcript without adding external knowledge.</think>Yes, Bill Gates discusses AI technologies in the provided context. He refers to AI as the "today's equivalent" of the opportunities he had in his youth, mentions the risks and significant potential it holds, and advises on how to use it for creative work and learning."""
```

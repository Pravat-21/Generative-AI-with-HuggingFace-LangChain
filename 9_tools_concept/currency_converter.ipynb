```python
import requests
from langchain_core.tools import tool   
```

```python
@tool
def currancy_bring(base_cur:str,target_cur:str)->float:
    """This function fetches the currency conversion factor between a given base currency and a target currency"""

    url=f"https://v6.exchangerate-api.com/v6/{secret_key_API}/pair/{base_cur}/{target_cur}"

    response=requests.get(url)

    return response.json()

@tool
def multiply(a:float,b:float)->float:
    """This fucntion can multiply with a & b and give us the result"""
    return a*b
```

```python
currancy_bring.invoke({'base_cur':'USD','target_cur':'INR'})
```

```python
from langchain_core.tools import InjectedToolArg
from typing import Annotated
from langchain_huggingface import ChatHuggingFace,HuggingFaceEndpoint
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
@tool
def convert(currency_value:float,conversion_rate:Annotated[float,InjectedToolArg])->float:
    """given a currency conversion rate this function calculates the target currency value from a given base currency value"""

    result=currency_value*conversion_rate
    return result
```

```python
llm=HuggingFaceEndpoint(
    repo_id="deepseek-ai/DeepSeek-V3.1", 
    task="text-generation"
    
)
model= ChatHuggingFace(llm=llm)
```

#### Model Binding

```python
model_with_tools=model.bind_tools([currancy_bring, convert, multiply])
```

```python
query="What is the conversion factor between INR and USD today "
```

```python
from langchain_core.messages import HumanMessage
```

```python
message=[]
message.append(HumanMessage(query))
```

```python
ai_message=model_with_tools.invoke(query)
```

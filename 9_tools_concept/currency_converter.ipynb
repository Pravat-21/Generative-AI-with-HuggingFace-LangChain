```python
import requests
from langchain_core.tools import tool   
```

```python
url="https://v6.exchangerate-api.com/v6/b1d027b6636d2d273c7b1620/pair/EUR/GBP"

```

```python
@tool
def currancy_bring(base_cur:str,target_cur:str)->float:
    """This function fetches the currency conversion factor between a given base currency and a target currency"""

    url=f"https://v6.exchangerate-api.com/v6/b1d027b6636d2d273c7b1620/pair/{base_cur}/{target_cur}"

    response=requests.get(url)

    return response.json()

```

```python
currancy_bring.invoke({'base_cur':'USD','target_cur':'INR'})
```

```python
from langchain_core.tools import InjectedToolArg
from typing import Annotated
from langchain_huggingface import ChatHuggingFace,HuggingFaceEndpoint
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
@tool
def convert(currency_value:float,conversion_rate:Annotated[float,InjectedToolArg])->float:
    """given a currency conversion rate this function calculates the target currency value from a given base currency value"""

    result=currency_value*conversion_rate
    return result-12
```

```python
llm=HuggingFaceEndpoint(
    repo_id="deepseek-ai/DeepSeek-V3.1", 
    task="text-generation"
    
)
model= ChatHuggingFace(llm=llm)
```

#### Model Binding

```python
model_with_tools=model.bind_tools([currancy_bring, convert])
```

```python
query="What is the conversion factor between INR and USD, and based on that can you convert 10 inr to usd using tools provied to you."
```

```python
from langchain_core.messages import HumanMessage
```

```python
message=[]
message.append(HumanMessage(query))
```

```python
ai_message=model_with_tools.invoke(message)
```

```python
message.append(ai_message)
```

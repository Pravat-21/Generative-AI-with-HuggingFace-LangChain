### Custom tools

#### 1. using **@Tool** decorator:

```python
from langchain_core.tools import tool
```

We need to create one fucntion with proper description & data type hints. Then just have to use the **tool** decorator.

```python
@tool
def sumation(a:int,b:int)->int:
    """ This takes the value of a & b then it return the samation of these value"""
    return a+b
```

```python
sumation.invoke({'a':8,'b':2})
```

```python
print(sumation.name)
print(sumation.description)
print(sumation.args)
```

#### 2. using StructuredTool & Pydantic:

```python
from langchain_core.tools import StructuredTool
from pydantic import Field,BaseModel
```

```python
## pydantic schema creation

class Schema(BaseModel):
    a :int=Field(required=True,description="This is the first number.")
    b :int=Field(required=True,description="This is the second number.")
```

```python
def multiply(a,b):
    return a*b
```

```python
mul_tool=StructuredTool.from_function(
    func=multiply,
    name="Multiply",
    description="This returns the multiplication bewteen a & b",
    args_schema=Schema

)
```

```python
mul_tool.invoke({'a':10,'b':7})
```

### 3. Using BaseTool:

```python
from langchain_core.tools import BaseTool
from typing import Type
```

```python
class MySchema(BaseModel):
    a :int=Field(required=True,description="This is the first number.")
    b :int=Field(required=True,description="This is the second number.")
```

```python
class SubstractTool(BaseTool):
    name : str = "Substract"
    description :str ="This tool give us the substaction of a & b"
    args_schema : Type[BaseModel] =MySchema

    def _run(self,a,b):
        return a-b
```

```python
my_tool=SubstractTool()
```

```python
my_tool.invoke({'a':5,'b':9})
```

## ToolKits

```python
@tool
def additon(a:int,b:int)->int:
    """This tool helps us by adding two numbers a,b"""
    return a+b
@tool
def multiply(a:int,b:int)->int:
    """This tool helps us by multiply two numbers a,b"""
    return a*b
```

```python
class MyToolKit:

    def tool_kit(self):
        return [additon,multiply]
```

```python
tools=MyToolKit()
```

```python
mytool=tools.tool_kit()
```

```python
for tool in mytool:
    print(tool.name,"&&&&&",tool.description)
```

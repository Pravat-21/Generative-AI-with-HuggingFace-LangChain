## AI-Agent

```python
from langchain_core.tools import tool
from langchain_community.tools import DuckDuckGoSearchRun
import requests
```

```python
search_tool=DuckDuckGoSearchRun()
```

```python
@tool
def get_weather_data(city:str)->str:
    """This fucntion fetchs the current weather data for a given city """
    url=f'https://api.weatherstack.com/current?access_key={secret_key of weather api}&query={city}'
    response=requests.get(url)
    return response.json()
```

```python
from langchain_huggingface import HuggingFaceEndpoint,ChatHuggingFace
```

```python
llm=HuggingFaceEndpoint(
    repo_id="deepseek-ai/DeepSeek-V3.1", 
    task="text-generation"
    
)
model= ChatHuggingFace(llm=llm)
```

```python
from langchain.agents import create_react_agent,AgentExecutor
from langchain import hub
```

```python
prompt = hub.pull('hwchase17/react')
```

```python
agent=create_react_agent(
    llm=model,
    tools=[search_tool, get_weather_data],
    prompt=prompt
)
```

```python
agent_executor=AgentExecutor(
    agent=agent,
    tools=[search_tool, get_weather_data],
    verbose=True
)
```

```python
response=agent_executor.invoke({"input":"Find the capital of west bengal, then find it's current weather condition"})
print(response)
```
